````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def show(title, img, cmap='gray'):
    plt.figure(figsize=(6,6))
    if img.ndim == 2:
        plt.imshow(img, cmap=cmap)
    else:
        plt.imshow(img)
    plt.title(title)
    plt.axis('off')
    plt.show()
````

````python
# Task 1 - Convolution

def convolve(img, kernel):
    """
    img    : 2D numpy array (grayscale)
    kernel : 2D numpy array
    """
    k_h, k_w = kernel.shape
    pad_h, pad_w = k_h // 2, k_w // 2

    # Pad image using reflect mode
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    H, W = img.shape
    output = np.zeros((H, W), dtype=np.float32)

    # Convolution operation
    for i in range(H):
        for j in range(W):
            region = padded[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output
````

````python
# Task 2 — Average Blur

# gray
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)
    return convolve(img, kernel)
# RGB
def average_blur_rgb(img, ksize):
    blurred = np.zeros_like(img, dtype=np.float32)

    for c in range(3):
        blurred[:, :, c] = average_blur_gray(img[:, :, c], ksize)

    return np.clip(blurred, 0, 255).astype(np.uint8)
````

````python
# Task 3 - Gaussian Blur

def gaussian_kernel(ksize, sigma):
    ax = np.arange(-ksize//2 + 1, ksize//2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)   # Normalize

    return kernel

# gray
def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

# RGB
def gaussian_blur_rgb(img, ksize, sigma):
    blurred = np.zeros_like(img, dtype=np.float32)

    for c in range(3):
        blurred[:, :, c] = gaussian_blur_gray(img[:, :, c], ksize, sigma)

    return np.clip(blurred, 0, 255).astype(np.uint8)

````

````python
def convolve(img, kernel):
    k_h, k_w = kernel.shape
    pad_h, pad_w = k_h // 2, k_w // 2

    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    H, W = img.shape
    output = np.zeros((H, W), dtype=np.float32)

    for i in range(H):
        for j in range(W):
            region = padded[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32)

# loding image
from PIL import Image

img = Image.open("uk_image.jpg").convert("RGB")
img = np.array(img)

# Convert to grayscale using average (no cv2)
gray = np.mean(img, axis=2).astype(np.float32)

show("Grayscale Image", gray)

# 1. & 2.
Gx = convolve(gray, sobel_x)
Gy = convolve(gray, sobel_y)

show("Sobel Gx", np.abs(Gx))
show("Sobel Gy", np.abs(Gy))

# 3.
grad_mag = np.sqrt(Gx**2 + Gy**2)

grad_mag_norm = (grad_mag / grad_mag.max()) * 255
grad_mag_norm = grad_mag_norm.astype(np.uint8)

show("Gradient Magnitude (Normalized)", grad_mag_norm)

# Thresholding keeps only strong edges
# Pixels below threshold are suppressed (set to 0)
threshold = 80
edge_map = np.zeros_like(grad_mag_norm)
edge_map[grad_mag_norm >= threshold] = 255

show("Thresholded Edge Map", edge_map)

Image.fromarray(edge_map).save("sobel_edge_map.png")
````

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def show(title, img, cmap='gray'):
    plt.figure(figsize=(6,6))
    plt.imshow(img, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.show()

def to_uint8(img_float):
    return np.clip(img_float, 0, 255).astype(np.uint8)

def convolve(img, kernel):
    k_h, k_w = kernel.shape
    pad_h, pad_w = k_h // 2, k_w // 2
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    H, W = img.shape
    out = np.zeros((H, W), dtype=np.float32)

    for i in range(H):
        for j in range(W):
            region = padded[i:i+k_h, j:j+k_w]
            out[i, j] = np.sum(region * kernel)
    return out

def gaussian_kernel(ksize, sigma):
    ax = np.arange(-ksize//2 + 1, ksize//2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    k = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    return k / np.sum(k)

def gaussian_blur_gray(img, ksize, sigma):
    k = gaussian_kernel(ksize, sigma)
    return convolve(img, k)

def unsharp_mask_gray(img, ksize=5, sigma=1.0, alpha=1.5):
    blurred = gaussian_blur_gray(img, ksize, sigma)
    mask = img - blurred
    sharp = img + alpha * mask
    return blurred, sharp

def magnitude_spectrum(img):
    """
    Magnitude spectrum visualization:
    - FFT -> shift zero frequency to center
    - log(1 + |F|) to compress huge range (otherwise mostly black)
    """
    F = np.fft.fft2(img)
    Fshift = np.fft.fftshift(F)

    mag = np.abs(Fshift)
    mag_log = np.log1p(mag)

    mag_norm = (mag_log / mag_log.max()) * 255
    return mag_norm.astype(np.uint8)

img_rgb = np.array(Image.open("uk_image.jpg").convert("RGB"))
gray = np.mean(img_rgb, axis=2).astype(np.float32)

show("Original Grayscale", gray)

gauss_ksize = 15   # bigger => stronger blur => stronger LPF
gauss_sigma = 3.0  # bigger => stronger blur => stronger LPF

unsharp_alpha = 2.0  # bigger => stronger sharpening => more HPF boost

blurred, sharp = unsharp_mask_gray(gray, ksize=gauss_ksize, sigma=gauss_sigma, alpha=unsharp_alpha)

show(f"Gaussian Blurred (ksize={gauss_ksize}, sigma={gauss_sigma})", to_uint8(blurred))
show(f"Unsharp Masked (alpha={unsharp_alpha})", to_uint8(sharp))

spec_orig   = magnitude_spectrum(gray)
spec_blur   = magnitude_spectrum(blurred)
spec_sharp  = magnitude_spectrum(sharp)

show("Magnitude Spectrum - Original", spec_orig)
show("Magnitude Spectrum - Gaussian Blurred (LPF)", spec_blur)
show("Magnitude Spectrum - Unsharp Masked (HPF Boost)", spec_sharp)

plt.figure(figsize=(12,8))

plt.subplot(2,3,1); plt.imshow(to_uint8(gray), cmap='gray'); plt.title("Original"); plt.axis('off')
plt.subplot(2,3,2); plt.imshow(to_uint8(blurred), cmap='gray'); plt.title("Gaussian Blurred"); plt.axis('off')
plt.subplot(2,3,3); plt.imshow(to_uint8(sharp), cmap='gray'); plt.title("Unsharp"); plt.axis('off')

plt.subplot(2,3,4); plt.imshow(spec_orig, cmap='gray'); plt.title("Spectrum: Original"); plt.axis('off')
plt.subplot(2,3,5); plt.imshow(spec_blur, cmap='gray'); plt.title("Spectrum: Blurred"); plt.axis('off')
plt.subplot(2,3,6); plt.imshow(spec_sharp, cmap='gray'); plt.title("Spectrum: Unsharp"); plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def show(title, img, cmap=None):
    plt.figure(figsize=(7,5))
    if img.ndim == 2:
        plt.imshow(img, cmap=cmap if cmap else 'gray')
    else:
        plt.imshow(img)
    plt.title(title)
    plt.axis("off")
    plt.show()

def to_uint8(x):
    return np.clip(x, 0, 255).astype(np.uint8)

def normalize_0_255(x):
    x = x.astype(np.float32)
    x = x - x.min()
    if x.max() > 1e-6:
        x = x / x.max()
    return (x * 255).astype(np.uint8)

# NumPy Convolution
def convolve(img, kernel):
    k_h, k_w = kernel.shape
    pad_h, pad_w = k_h // 2, k_w // 2

    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    H, W = img.shape
    out = np.zeros((H, W), dtype=np.float32)

    for i in range(H):
        for j in range(W):
            region = padded[i:i+k_h, j:j+k_w]
            out[i, j] = np.sum(region * kernel)
    return out

# Load image
try:
    img = img_part5.copy()
except NameError:
    img = np.array(Image.open("uk_image.jpg").convert("RGB"))

img = img.astype(np.float32)

show("Original Image (Part E)", img.astype(np.uint8))
````

````python
# 1. Analyze RGB values

def mean_rgb_in_box(img, x1, y1, x2, y2):
    patch = img[y1:y2, x1:x2, :]
    return patch.reshape(-1,3).mean(axis=0)

H, W = img.shape[:2]

# yellow leaf area
yellow_box = (int(0.55*W), int(0.15*H), int(0.75*W), int(0.35*H))
# green leaf area
green_box  = (int(0.10*W), int(0.25*H), int(0.30*W), int(0.45*H))

yr, yg, yb = mean_rgb_in_box(img, *yellow_box)
gr, gg, gb = mean_rgb_in_box(img, *green_box)

print("Avg RGB (Yellow box):", (yr, yg, yb))
print("Avg RGB (Green box) :", (gr, gg, gb))

# 2. Build a “Yellow-ish” Mask

R = img[:,:,0]
G = img[:,:,1]
B = img[:,:,2]

# "Yellowness" score: yellow has high (R+G) and low B
yellowness = (R + G) / 2 - B

# Make a binary mask using thresholds
thr_y = np.percentile(yellowness, 75)   # top 25% yellowness
thr_b = np.percentile(B, 60)            # B should be relatively low

yellow_mask = (yellowness >= thr_y) & (B <= thr_b)
yellow_mask_u8 = (yellow_mask.astype(np.uint8) * 255)

show("Yellow-ish Mask (binary)", yellow_mask_u8, cmap="gray")

# 3. Custom colour edge response using (R − G) gradient

diff_RG = (R - G).astype(np.float32)
show("R - G (visualized)", normalize_0_255(diff_RG), cmap="gray")

kx = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

ky = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32)

gx = convolve(diff_RG, kx)
gy = convolve(diff_RG, ky)

edge_strength = np.sqrt(gx**2 + gy**2)
edge_norm = normalize_0_255(edge_strength)

show("Edge Strength on (R-G) [custom colour edge]", edge_norm, cmap="gray")

# 4. Combine response maps + get clean outline around yellow leaf

# Expand the yellow mask using convolution with a box (cheap dilation)
# This grows the mask so boundary pixels around yellow are included.
grow_kernel = np.ones((9,9), dtype=np.float32)
grown = convolve(yellow_mask.astype(np.float32), grow_kernel)
yellow_mask_grown = (grown > 0).astype(np.uint8)  # any nearby yellow -> keep
yellow_mask_grown_u8 = yellow_mask_grown * 255

show("Yellow Mask (grown)", yellow_mask_grown_u8, cmap="gray")

# Keep only edges near the yellow region
yellow_edges = edge_norm * yellow_mask_grown
show("Edges near Yellow Region (before threshold)", yellow_edges, cmap="gray")

# Threshold for clean outline
t = np.percentile(yellow_edges[yellow_edges > 0], 70) if np.any(yellow_edges > 0) else 0
outline = (yellow_edges >= t).astype(np.uint8) * 255

show("Final Clean Outline (thresholded)", outline, cmap="gray")

overlay = img.astype(np.uint8).copy()
# draw outline in red
overlay[outline > 0] = [255, 0, 0]
show("Outline Overlay (Red on Original)", overlay)

````

````python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image

def show(title, img):
    plt.figure(figsize=(7,5))
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")
    plt.show()

# Load image
try:
    img_rgb = img_part5.copy()
except NameError:
    img_rgb = np.array(Image.open("uk_image.jpg").convert("RGB"))

show("Original", img_rgb)

# cv2 expects BGR, so convert RGB -> BGR first
img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

# Tune these if mask misses parts:
lower_yellow = np.array([18, 80, 80])   # H, S, V
upper_yellow = np.array([40, 255, 255])

mask = cv2.inRange(hsv, lower_yellow, upper_yellow)

# Clean mask (optional but gives better result)
kernel = np.ones((5,5), np.uint8)
mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=2)
mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=1)

plt.figure(figsize=(6,4))
plt.imshow(mask, cmap="gray")
plt.title("Yellow Mask")
plt.axis("off")
plt.show()

hsv_red1 = hsv.copy()

# Set Hue to red. In OpenCV: red is near H=0 (also near 179 wrap-around)
hsv_red1[..., 0][mask > 0] = 0      # Hue -> 0 (red)

# Keep saturation/value same (natural shading preserved)
out_bgr1 = cv2.cvtColor(hsv_red1, cv2.COLOR_HSV2BGR)
out_rgb1 = cv2.cvtColor(out_bgr1, cv2.COLOR_BGR2RGB)

show("Yellow → Bright Red", out_rgb1)

hsv_red2 = hsv.copy()
hsv_red2[..., 0][mask > 0] = 0      # Hue -> red

# Make it slightly less bright by reducing V on masked region
hsv_red2[..., 2][mask > 0] = np.clip(hsv_red2[..., 2][mask > 0] * 0.85, 0, 255).astype(np.uint8)

out_bgr2 = cv2.cvtColor(hsv_red2, cv2.COLOR_HSV2BGR)
out_rgb2 = cv2.cvtColor(out_bgr2, cv2.COLOR_BGR2RGB)

show("Yellow → Darker Red", out_rgb2)

````
