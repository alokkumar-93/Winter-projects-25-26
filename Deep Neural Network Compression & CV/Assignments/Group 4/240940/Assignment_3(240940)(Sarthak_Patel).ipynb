````python
# Converting the image to a grayscale image
R = img_rgb[:, :, 0].astype(np.float32)
G = img_rgb[:, :, 1].astype(np.float32)
B = img_rgb[:, :, 2].astype(np.float32)
gray_numpy = (0.299 * R + 0.587 * G + 0.114 * B).astype(np.uint8)
plt.figure(figsize=(6,6))
plt.imshow(gray_numpy,cmap="gray")
plt.axis("off")
plt.title("Original Image in Grayscale Numpy")
plt.show()
````

````python
# writing the convolve function.
def convolve(img, kernel):
    kernel_height, kernel_width = kernel.shape
    pad_height = (kernel_height - 1) // 2
    pad_width = (kernel_width - 1) // 2

    padded_img = np.pad(img,
                        pad_width=((pad_height, pad_height),
                                   (pad_width, pad_width)),
                        mode='reflect')

    img_height, img_width = img.shape
    output = np.zeros_like(img, dtype=np.float64)

    for i in range(img_height):
        for j in range(img_width):
            patch = padded_img[i:i + kernel_height,
                              j:j + kernel_width]
            output[i, j] = np.sum(patch * kernel)

    return output
````

````python
# Average Blur for Grayscale.
def average_blur_gray(img, ksize):

   # kernel size should be odd

    box_kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)
    output = convolve(img, box_kernel)
    return output

````

````python
# Average Blur for RGB Images.
def average_blur_rgb(img_rgb, ksize):


    if img_rgb.ndim != 3 or img_rgb.shape[-1] != 3:
        raise ValueError("Image must be 3D RGB")
    if ksize % 2 == 0:
        raise ValueError("Kernel size must be odd")

    # Converting the image to float 32 values.
    img_float = img_rgb.astype(np.float32)

    # Creating the kernel
    box_kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)
    output = np.zeros_like(img_float)

    for channel in range(3):
        output[:, :, channel] = convolve(img_float[:, :, channel], box_kernel)

    return np.clip(output, 0, 255).astype(np.uint8)
````

````python
# Gaussian Kernal Generator.
def gaussian_kernel(ksize, sigma):

    if ksize % 2 == 0:
        raise ValueError("Kernel size must be odd")
    if sigma <= 0:
        raise ValueError("Sigma must be positive")

    ax = np.arange(-ksize // 2 + 1, ksize // 2 + 1, dtype=np.float32)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)
    return kernel
# Gaussian Blur for Grayscale Images.
def gaussian_blur_gray(img, ksize, sigma):

    kernel = gaussian_kernel(ksize, sigma)
    output = convolve(img, kernel)
    return output
# Gaussian Blur for RGB Images.
def gaussian_blur_rgb(img_rgb, ksize, sigma):

    if img_rgb.ndim != 3 or img_rgb.shape[-1] != 3:
        raise ValueError("Image must be 3D RGB")

    # Validating the parameters
    if ksize % 2 == 0:
        raise ValueError("Kernel size must be odd")
    if sigma <= 0:
        raise ValueError("Sigma must be positive")

    kernel = gaussian_kernel(ksize, sigma)


    img_float = img_rgb.astype(np.float32)
    output = np.zeros_like(img_float)


    for channel in range(3):
        output[:, :, channel] = convolve(img_float[:, :, channel], kernel)


    return np.clip(output, 0, 255).astype(np.uint8)
````

````python
# 1. Magnitude spectrum
#Center= Represents low frequencies (smooth areas, background).
#Corners= Represent high frequencies (edges, texture, noise).

# 2. Gaussian blur
# The spectrum becomes noticeably darker at the outer edges because Gaussian convolution averages pixels, which destroys fine details.
# In the frequency domain, this acts as a "low-pass filter" that attenuates high-frequency components (outer regions) while preserving the low-frequency  center.
# Increasing 'ksize' or 'sigma' narrows the Gaussian curve in the frequency domain.
# This makes the bright central spot smaller (more high frequencies cut off), resulting in a blurrier image.

# 3. Unsharp Masking
# The outer regions (high frequencies) become brighter compared to the original.
# Unsharp masking adds the difference or edges back to the original.
# Sharpened = Original + alpha * high frequencies
# This amplifies the high-frequency components relative to the low ones.

#  Alpha values:
# Increasing 'alpha' directly boosts the amplitude of high frequencies.
# Visually, the outer spectrum gets brighter.


# 4. Conclusion:
#  Convolution in spatial domain = Multiplication in frequency domain
#  A broad  kernel (large blur) is a narrow frequency filter (cuts many frequencies).
#  A sharp delta like identity passes all frequencies.
#  By manipulating the kernel, we directly control which frequencies pass through to the final image.
````

````python
# Sobel matrices
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]], dtype=np.float32)
sobel_y = np.array([[-1,-2,-1], [0, 0, 0], [1, 2, 1]], dtype=np.float32)

rg_kernel_x = np.array([[-2, 0, 2], [-4, 0, 4], [-2, 0, 2]], dtype=np.float32)
rg_kernel_y = np.array([[-2,-4,-2], [0, 0, 0], [2, 4, 2]], dtype=np.float32)
````

````python
# computing red green difference and then edge magnitudes for r,g and r-g using sobel.
r_channel = img_part5[:, :, 0].astype(np.float32)
g_channel = img_part5[:, :, 1].astype(np.float32)
rg_diff = r_channel - g_channel

r_gx = ndimage.convolve(r_channel, sobel_x, mode='reflect')
r_gy = ndimage.convolve(r_channel, sobel_y, mode='reflect')
r_mag = np.sqrt(r_gx**2 + r_gy**2)

g_gx = ndimage.convolve(g_channel, sobel_x, mode='reflect')
g_gy = ndimage.convolve(g_channel, sobel_y, mode='reflect')
g_mag = np.sqrt(g_gx**2 + g_gy**2)

rg_gx = ndimage.convolve(rg_diff, rg_kernel_x, mode='reflect')
rg_gy = ndimage.convolve(rg_diff, rg_kernel_y, mode='reflect')
rg_mag = np.sqrt(rg_gx**2 + rg_gy**2)

````

````python
# Assigning weights to each map and then linearly combining the magnitude maps using the weights.
w_r = 0.3      # Red: low-medium
w_g = 0.3      # Green: low-medium
w_rg = 2.0     # R-G Highest.

print("Weights: R={}, G={}, R-G={}".format(w_r, w_g, w_rg))

combined = w_r * r_mag + w_g * g_mag + w_rg * rg_mag
print("Combined: min={:.1f}, max={:.1f}".format(
    combined.min(), combined.max()))

combined = w_r * r_mag + w_g * g_mag + w_rg * rg_mag
print("Combined: min={:.1f}, max={:.1f}".format(
    combined.min(), combined.max()))

combined_edges = (w_r * r_mag +
                  w_g * g_mag +
                  w_rg * rg_mag)


print("  Min: {:.1f}".format(combined_edges.min()))
print("  Max: {:.1f}".format(combined_edges.max()))
print("  Mean: {:.1f}".format(combined_edges.mean()))
````

````python
# Step 1: Normalize to [0, 255]
combined_norm = (combined_edges - combined_edges.min()) / \
                (combined_edges.max() - combined_edges.min() + 1e-6) * 255


# List of threshold values.
thresholds = [40, 50, 60, 70, 80, 90, 100]

for thresh in thresholds:
    binary_test = (combined_norm > thresh).astype(np.uint8) * 255
    pixels = np.sum(binary_test > 0) # Counting the total no of white pixels (detected edges)
    percent = 100 * pixels / binary_test.size

    if thresh == 70: # we chode this threshold as it seenms to be the best balance between dteecting edges and avoiding noise.
        status = "OPTIMAL"
    else:
        status = ""

    print("  Thresh {:3d}: {:6d} pixels ({:5.2f}%) {}".format(
        thresh, pixels, percent, status))
````

````python
optimal_threshold = 70
final_edges = (combined_norm > optimal_threshold).astype(np.uint8) * 255


print("  Edge pixels: {}".format(np.sum(final_edges > 0)))
print("  Percentage: {:.2f}%".format(100 * np.sum(final_edges > 0) / final_edges.size))
````

````python
def rgb_to_hsv(img_rgb):

  img_float = img_rgb.astype(np.float32) / 255.0

  # Extracting rg and b channels
  r = img_float[:,:,0]
  g = img_float[:,:,1]
  b = img_float[:,:,2]

  # calculating cmax,cmin,delta
  cmax = np.max(img_float, axis=2)
  cmin = np.min(img_float, axis=2)
  delta = cmax - cmin

  # Initializing HSV channels
  h = np.zeros_like(cmax)
  s = np.zeros_like(cmax)
  v = cmax


  # Identifying pixels where delta is not zero to avoid division by zero
  non_zero_delta = delta != 0

  # Conditions for hue calculation based on which channel is Cmax
  cond_r = (cmax == r) & non_zero_delta
  cond_g = (cmax == g) & non_zero_delta
  cond_b = (cmax == b) & non_zero_delta

  # Apply hue calculation based on conditions
  h[cond_r] = (((g[cond_r] - b[cond_r]) / delta[cond_r])) % 6
  h[cond_g] = (((b[cond_g] - r[cond_g]) / delta[cond_g]) + 2)
  h[cond_b] = (((r[cond_b] - g[cond_b]) / delta[cond_b]) + 4)

  h = h * 60.0
  h[h < 0] += 360 # This step is to ensure that the hue is positive ie this converts any negative value to positive value by adding 360 degrees.

  h = h / 2.0

  # Saturation
  # Identifying pixels where Cmax is not zero to avoid division by zero
  non_zero_v = cmax != 0
  s[non_zero_v] = (delta[non_zero_v] / cmax[non_zero_v]) * 255

  v = v * 255

  h = np.clip(h, 0, 179).astype(np.uint8)
  s = np.clip(s, 0, 255).astype(np.uint8)
  v = np.clip(v, 0, 255).astype(np.uint8)

  # Stack the H, S, V channels to form an HSV image array
  hsv_image_np = np.dstack([h, s, v])

  return hsv_image_np


hsv_output = rgb_to_hsv(img_part5)


plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img_part5)
plt.axis("off")
plt.title("Original Image (RGB)")

plt.subplot(1, 2, 2)
plt.imshow(hsv_output)
plt.axis("off")
plt.title("HSV Image.")
plt.show()
````
