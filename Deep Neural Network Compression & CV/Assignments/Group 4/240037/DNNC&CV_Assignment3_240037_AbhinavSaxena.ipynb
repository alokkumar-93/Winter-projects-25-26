````python
import numpy as np
import matplotlib.pyplot as plt
import requests
from PIL import Image
from io import BytesIO

# Fetch the image from Picsum API
url = "https://picsum.photos/200"
response = requests.get(url)

# Load image with PIL
img = Image.open(BytesIO(response.content)).convert("RGB")

# Convert to numpy array
img_np = np.array(img)

# Convert to grayscale (manual, human-style)
gray = (0.299 * img_np[:,:,0] +
        0.587 * img_np[:,:,1] +
        0.114 * img_np[:,:,2]).astype(np.float32)
````

````python
def convolve(img, kernel):
    """
    img    : 2D numpy array (grayscale image)
    kernel : 2D numpy array
    returns: convolved image
    """

    kh, kw = kernel.shape
    pad_h = kh // 2
    pad_w = kw // 2

    # Reflect padding
    padded = np.pad(img,
                    ((pad_h, pad_h), (pad_w, pad_w)),
                    mode='reflect')

    h, w = img.shape
    output = np.zeros((h, w), dtype=np.float32)

    # Flip kernel for convolution
    kernel_flipped = np.flipud(np.fliplr(kernel))

    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel_flipped)

    return output

def average_blur_gray(img, ksize):
    """
    img   : grayscale image (2D)
    ksize : kernel size (odd integer)
    """

    kernel = np.ones((ksize, ksize), dtype=np.float32)
    kernel = kernel / (ksize * ksize)

    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    """
    img : RGB image (H, W, 3)
    """

    blurred = np.zeros_like(img, dtype=np.float32)

    for c in range(3):
        blurred[:,:,c] = average_blur_gray(img[:,:,c], ksize)

    return blurred.astype(np.uint8)
def gaussian_kernel(ksize, sigma):
    """
    Returns a normalized 2D Gaussian kernel
    """

    center = ksize // 2
    kernel = np.zeros((ksize, ksize), dtype=np.float32)

    for i in range(ksize):
        for j in range(ksize):
            x = i - center
            y = j - center
            kernel[i, j] = np.exp(-(x**2 + y**2) / (2 * sigma**2))

    # Normalize so sum = 1
    kernel = kernel / np.sum(kernel)

    return kernel
def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)


def gaussian_blur_rgb(img, ksize, sigma):
    blurred = np.zeros_like(img, dtype=np.float32)

    for c in range(3):
        blurred[:,:,c] = gaussian_blur_gray(img[:,:,c], ksize, sigma)

    return blurred.astype(np.uint8)

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)


def gaussian_blur_rgb(img, ksize, sigma):
    blurred = np.zeros_like(img, dtype=np.float32)

    for c in range(3):
        blurred[:,:,c] = gaussian_blur_gray(img[:,:,c], ksize, sigma)

    return blurred.astype(np.uint8)

avg_gray = average_blur_gray(gray, 5)
avg_rgb = average_blur_rgb(img_np, 5)

gauss_gray = gaussian_blur_gray(gray, 5, 1.0)
gauss_rgb = gaussian_blur_rgb(img_np, 5, 1.0)

plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.title("Original RGB")
plt.imshow(img_np)
plt.axis("off")

plt.subplot(2,3,2)
plt.title("Average Blur RGB")
plt.imshow(avg_rgb)
plt.axis("off")

plt.subplot(2,3,3)
plt.title("Gaussian Blur RGB")
plt.imshow(gauss_rgb)
plt.axis("off")

plt.subplot(2,3,4)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2,3,5)
plt.title("Average Blur Gray")
plt.imshow(avg_gray, cmap='gray')
plt.axis("off")

plt.subplot(2,3,6)
plt.title("Gaussian Blur Gray")
plt.imshow(gauss_gray, cmap='gray')
plt.axis("off")

plt.show()


````

````python

sobel_x = np.array([[-1,  0,  1],
                    [-2,  0,  2],
                    [-1,  0,  1]], dtype=np.float32)

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]], dtype=np.float32)
# Apply Sobel filters
Gx = convolve(gray, sobel_x)
Gy = convolve(gray, sobel_y)
# Gradient magnitude
gradient_magnitude = np.sqrt(Gx**2 + Gy**2)
# Normalize gradient magnitude to range 0–255
grad_min = gradient_magnitude.min()
grad_max = gradient_magnitude.max()

edge_map = (gradient_magnitude - grad_min) / (grad_max - grad_min)
edge_map = (edge_map * 255).astype(np.uint8)

# Optional thresholding to get stronger edges only
# Any value below threshold is suppressed (set to 0)

threshold = 100
edge_thresh = edge_map.copy()
edge_thresh[edge_thresh < threshold] = 0

plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2,3,2)
plt.title("Sobel Gx")
plt.imshow(np.abs(Gx), cmap='gray')
plt.axis("off")

plt.subplot(2,3,3)
plt.title("Sobel Gy")
plt.imshow(np.abs(Gy), cmap='gray')
plt.axis("off")

plt.subplot(2,3,4)
plt.title("Gradient Magnitude")
plt.imshow(edge_map, cmap='gray')
plt.axis("off")

plt.subplot(2,3,5)
plt.title("Thresholded Edge Map")
plt.imshow(edge_thresh, cmap='gray')
plt.axis("off")

plt.show()

````

#### Task - Frequency Analysis (LPF & HPF Verification)
Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
````

````python
def magnitude_spectrum(img):

    # Compute 2D FFT
    f = np.fft.fft2(img)

    # Shift zero frequency to center
    fshift = np.fft.fftshift(f)

    # Magnitude spectrum (log for better visualization)
    magnitude = np.log(1 + np.abs(fshift))

    return magnitude

# Original image spectrum
mag_original = magnitude_spectrum(gray)

# Gaussian blurred image spectrum (LPF)
mag_blur = magnitude_spectrum(blurred)

# Unsharp masked image spectrum (HPF effect)
mag_unsharp = magnitude_spectrum(unsharp)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original Image Spectrum")
plt.imshow(mag_original, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Gaussian Blurred Spectrum (LPF)")
plt.imshow(mag_blur, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Unsharp Masking Spectrum (HPF)")
plt.imshow(mag_unsharp, cmap='gray')
plt.axis("off")

plt.show()

````

````python
# Stronger Gaussian blur (more low-pass filtering)
blur_strong = gaussian_blur_gray(gray, ksize=9, sigma=2.5)
mag_blur_strong = magnitude_spectrum(blur_strong)

# Stronger sharpening (higher alpha)
alpha_high = 1.5
mask_high = gray - blur_strong
unsharp_high = gray + alpha_high * mask_high
unsharp_high = np.clip(unsharp_high, 0, 255)

mag_unsharp_high = magnitude_spectrum(unsharp_high)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Strong Gaussian Blur Spectrum")
plt.imshow(mag_blur_strong, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Strong Unsharp Masking Spectrum")
plt.imshow(mag_unsharp_high, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Original Spectrum (Reference)")
plt.imshow(mag_original, cmap='gray')
plt.axis("off")

plt.show()

````

````python
# In magnitude spectrum, center part is bright which means low frequencies are present there.
# After applying Gaussian blur, image becomes smooth and in spectrum the outer area becomes darker.
# This shows that blur is removing high frequency details like edges. Increasing sigma makes this effect stronger.
# In unsharp masking, when I increase alpha, edges become sharper and in spectrum the outer region becomes brighter,meaning high frequencies are increased.
# For negative alpha, image looks blurred and spectrum is similar to Gaussian blur case.
#So basically, Gaussian blur works as low pass filter and sharpening boosts high frequencies, which confirms how convolution filters frequencies.
````
