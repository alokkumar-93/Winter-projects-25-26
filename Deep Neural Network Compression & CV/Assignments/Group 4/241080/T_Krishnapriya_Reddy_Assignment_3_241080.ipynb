````python
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np
img=Image.open("image.jpg.jpg").convert("L")
gray_img=np.array(img)
img_rgb=Image.open("image.jpg.jpg").convert("RGB")
rgb_img=np.array(img_rgb)
kernel=np.array([[0,-1,0],[-1,4,-1],[0,-1,0]])
````

````python
def convolve(gray_img,kernel):
  gray_img=gray_img.astype(np.float64)
  kh,kw=kernel.shape
  pad_h=kh//2
  pad_w=kw//2
  padded_img=np.pad(gray_img,((pad_h,pad_h),(pad_w,pad_w)),mode='reflect')
  H,W=gray_img.shape
  output=np.zeros((H,W))
  for i in range(H):
      for j in range(W):
          region=padded_img[i:i+kh,j:j+kw]
          output[i,j]=np.sum(region*kernel)
  return output
convolved_img=convolve(gray_img,kernel)
plt.title("Original")
plt.imshow(rgb_img)
plt.axis("off")
plt.show()
plt.imshow(convolved_img,cmap='gray')
plt.title("Convolved image gray")
plt.axis("off")
plt.show()
````

````python
def average_blur_gray(gray_img,ksize):
  kernel=np.ones((ksize,ksize))/(ksize*ksize)
  return convolve(gray_img,kernel)
plt.title("Average Blur (Gray)")
plt.imshow(average_blur_gray(gray_img,9),cmap='gray')
plt.axis("off")
plt.show()
````

````python
def average_blur_rgb(img,ksize):
  blurred=np.zeros_like(img,dtype=np.float64)
  for c in range(3):
    blurred[:,:,c]=average_blur_gray(rgb_img[:,:,c],ksize)
  return blurred
plt.title("Average Blur (RGB)")
plt.imshow(average_blur_rgb(rgb_img,9).astype(np.uint8))
plt.axis("off")
plt.show()
````

````python
def gaussian_kernel(ksize,sigma):
  ax=np.arange(-(ksize//2),(ksize//2)+1)
  xx,yy=np.meshgrid(ax,ax)
  kernel=np.exp(-(xx**2+yy**2)/(2*sigma**2))
  kernel/=np.sum(kernel)
  return kernel
````

````python
def gaussian_blur_gray(gray_img,ksize,sigma):
  kernel=gaussian_kernel(ksize,sigma)
  return convolve(gray_img,kernel)
plt.title("Gaussian Blur (Gray)")
plt.imshow(gaussian_blur_gray(gray_img,9,1),cmap='gray')
plt.axis("off")
````

````python
def gaussian_blur_rgb(rgb_img,ksize,sigma):
  blurred=np.zeros_like(rgb_img,dtype=np.float64)
  for c in range(3):
      blurred[:,:,c]=gaussian_blur_gray(rgb_img[:,:,c],ksize,sigma)
  return blurred
plt.title("Gaussian Blur (RGB)")
plt.imshow(gaussian_blur_rgb(rgb_img,9,5).astype(np.uint8))
plt.axis("off")
````

````python
def magnitude_spectrum(img):
    img=img.astype(np.float64)
    F=np.fft.fft2(img)
    F_shift=np.fft.fftshift(F)
    magnitude=np.log(1+np.abs(F_shift))
    return magnitude
plt.title("Original")
plt.imshow(gray_img,cmap='gray')
plt.axis("off")
plt.show()
blurred=gaussian_blur_gray(gray_img,ksize=9,sigma=2)
plt.title("Gaussian blurred")
plt.imshow(blurred,cmap='gray')
plt.axis("off")
plt.show()
````

````python
R=img_part5[:,:,0].astype(np.float64)
G=img_part5[:,:,1].astype(np.float64)
B=img_part5[:,:,2].astype(np.float64)
RG_diff=R-G
edge_kernel=np.array([[-1,-1,-1],[-1,8,-1],[-1,-1,-1]])
edge_RG=convolve(RG_diff,edge_kernel)
edge_R=convolve(R,edge_kernel)
edge_G=convolve(G,edge_kernel)
edge_B=convolve(B,edge_kernel)
combined_edges=edge_RG+0.5*edge_R+0.5*edge_G-0.5*edge_B
combined_edges=np.abs(combined_edges)
combined_edges=combined_edges/combined_edges.max()
combined_edges=(combined_edges*255).astype(np.uint8)
threshold=80
edge_map=np.zeros_like(combined_edges)
edge_map[combined_edges>threshold]=255
plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(img_part5)
plt.axis("off")
plt.subplot(1,3,2)
plt.title("Colour edge response")
plt.imshow(combined_edges,cmap='gray')
plt.axis("off")
plt.subplot(1,3,3)
plt.title("Yellow leaf edge map")
plt.imshow(edge_map,cmap='gray')
plt.axis("off")
plt.show()
````
